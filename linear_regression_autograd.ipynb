```python
import torch
```

```python
import numpy as np
```

```python
in1 = np.arange(1.0, 11.0)
print(in1)
```

```python
X = torch.from_numpy(in1)
print(X)
print(X.shape)
print(X.dtype)
```

```python
out1 = np.arange(2.0, 21.0, step=2)
print(out1)
Y = torch.from_numpy(out1)
print(Y)
print(Y.shape)
print(Y.dtype)
```

```python
w = torch.tensor(1, dtype=torch.float32, requires_grad=True)
print(w)
```

```python
print(w.grad)
```

```python
# Model output
def forward(x):
  return w * x

# loss = Mean Squared Error (MSE)
def loss(y, y_pred):
  return ((y_pred-y)**2).mean()

# Evaluation
X_test = [0.5, 32, 200]
```

```python
print(f"Prediction before training: f(X_test):")
for item in X_test:
  y_pred = forward(item).item()
  print(y_pred)
```

```python
# Training
w = torch.tensor(0, dtype=torch.float32, requires_grad=True)
print(w)
learning_rate = 0.01
epochs = 20

for epoch in range(epochs):
  # print(f"epoch = {epoch}")
  y_pred = forward(X)
  # print(f"y, y_pred = {Y}, {y_pred}")

  l = loss(Y, y_pred)
  # print(f"loss = {l}")

  # Calculate gradient = backward pass
  l.backward()
  # print(f"backward = {l}")

  # Update weights
  with torch.no_grad():
    w -= (learning_rate * w.grad)
    # print(f"w = {w}")
    # print(f"w.dtype = {w.dtype}")
    # print(f"w.grad = {w.grad}")

  # Zero the gradients after updating
  w.grad.zero_()

  # if (epoch+1) % 10 == 0:
  print(f"epoch ({epoch+1}); w ({w.item()}); loss ({l.item()})")
  # break


```

```python
print(f"Prediction after training: f(X_test):")
for item in X_test:
  y_pred = forward(item).item()
  print(y_pred)
```
