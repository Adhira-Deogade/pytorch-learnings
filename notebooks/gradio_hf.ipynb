```python
from gradio_client import Client

client = Client("black-forest-labs/FLUX.1-schnell")
result = client.predict(
		prompt="A serene landscape of colonial American architecture, featuring well-preserved buildings with wooden facades and gabled roofs. The setting should include rolling hills and a clear blue sky, evoking a sense of determination and progress",
		seed=0,
		randomize_seed=True,
		width=1024,
		height=1024,
		num_inference_steps=4,
		api_name="/infer"
)
print(result)
image = Image.open(result[0]).convert("RGB")
image.save(f"{result[0].replace('webp', 'png')}", format="png")
```
