```python
# Pytorch autograd package provides automatic differentiation
# for all operations on tensors
# torch.autograd is an engine which provides/computes - partial differentiation
# applying the chain rule
```

```python
import torch
```

```python
x = torch.rand(5,3, requires_grad=True)
print(x)
```

```python
y = x + 2
print(y)
```

```python
print(x-y)
```

```python
# y was created as a function of some operation, so it has grad_fn attribute
print(y.grad_fn)
```

```python
print(x.grad_fn)
```

```python
# Do more operations
z = x * y * 6
print(z)
print(z.mean())
```

```python
# Let's compute the gradients with backpropagation
# We perform a computation, once it's completed, we can call .backwards()
# This will compute dervation for us
```

```python
# grad can be implicitly created only for scalar outputs
print(x.grad)
z.mean().backward()
print(x.grad) # dz/dx
```

```python
# All the operations will be tracked, when we set requires_grad=True
# We are going through a forward pass in our network
```

```python
# After the forward pass, in the end we calculate the loss, mean squared error
# Then we apply backpropagation on loss, which computes derivative
# of loss with respect to weights
```

```python
# After each iteration, we need to empty the gradient with optimizer.zero_grad()
```

```python
# Stop a tensor from tracking history
```

```python
a = torch.rand(5,3)
print(a)
b = (a * a).sum()
print(a.requires_grad)
print(b.grad_fn)
```

```python
a = torch.rand(5,3, requires_grad=True)
print(a)
b = (a * a).sum()
print(a.requires_grad)
print(b.grad_fn)
print(b.grad)
```

```python
a = torch.rand(5,3, requires_grad=True)
print(a)
b = a.detach() # This will create a copy
print(a.requires_grad)
print(b.requires_grad)
```

```python
a = torch.rand(5,3, requires_grad=True)
print(a)
with torch.no_grad():
    b = a * 2
    print(b)
print(a.requires_grad)
print(b.requires_grad)
```

```python
# The above step is required when we are in evaluation stage
```
