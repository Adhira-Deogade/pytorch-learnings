```python
# A typical pytorch pipeline looks like this -
# 1. Design - inputs, outputs, forward pass through layers
# 2. Construct - loss and optimizer
# 3. Training loop
#   - forward pass - compute prediction and loss
#   - compute gradient
#   - update weights
```

```python
# Linear regression example in the pytorch way
```

```python
import torch
import torch.nn as nn
```

```python
# Linear regression
# f(x) =  wx
# here: f(x) = 2x
```

```python
import numpy as np
```

```python
in1 = np.arange(1,10, dtype=np.float32)
print(in1)
```

```python
in1_torch = torch.from_numpy(in1)
print(in1_torch)
```

```python
in1_torch.shape
```

```python
# Now, I want a 9 by 1 tensor, how will I achieve that?
# Let's try reshaping
```

```python
X = in1_torch.view(9,1)
```

```python
print(X.shape)
```

```python
print(X)
```

```python
Y = 2 * X # This creates a copy of tensor, nothing changes in memory
```

```python
print(Y)
```

```python
n_samples, n_features = X.shape
print(n_samples, n_features)
```

```python
# Create a tensor for evaluation/ testing
in_test = np.arange(11,20, dtype=np.float32)
in_test_torch = torch.from_numpy(in_test)
X_test = in_test_torch.view(9,1)
# X_test = torch.tensor([5], dtype=torch.float32)
```

```python
print(X_test)
```

```python
X.dtype
```

```python
Y.dtype
```

```python
# 1. Design model, the model has to implement forward pass
# We can use built-in model from pytorch
# model = nn.Linear(input_size, output_size)
```

```python
class LinearRegression(nn.Module):
  def __init__(self, input_dim, output_dim):
    # Define all the layers
    super(LinearRegression, self).__init__()
    # We are creating only 1 layer here
    self.lin = nn.Linear(input_dim, output_dim)

  def forward(self, x):
    # Get the input tensor
    # Apply all the layers
    return self.lin(x)
```

```python
input_size, output_size = n_features, n_features
```

```python
print(input_size, output_size)
```

```python
linear_model = LinearRegression(input_dim=input_size, output_dim=output_size)
```

```python
print(linear_model)
```

```python
print(f"Prediction before training: f({X_test}):")
print(linear_model(X_test))
```

```python
# 2. Define loss and optimizer
learning_rate = 0.01
epochs = 100
loss = nn.MSELoss()
optimizer = torch.optim.SGD(linear_model.parameters(), lr=learning_rate)

print(f"learning rate: {learning_rate}")
print(f"epochs: {epochs}")
print(f"loss: {loss}")
print(f"optimizer: {optimizer}")
```

```python
# 3. Training loop
for epoch in range(epochs):
  # Make a prediction
  y_pred = linear_model.forward(X)

  # Compute loss
  l = loss(Y, y_pred)

  # Compute gradient descent
  l.backward() # dl/dw

  # Update weights
  optimizer.step()

  # Zero the gradient for next iteration
  optimizer.zero_grad()

  if (epoch+1) % 10 == 0:
    weight, bias = linear_model.parameters()
    print(f"epoch({epoch+1}); loss({l}); weight({weight[0][0]}); bias({bias[0]})")
```

```python
print(f"Prediction after training: f({X_test}): {linear_model(X_test)}")
```

```python
print(f"Prediction after training: f(100): {linear_model(torch.tensor([100.0]))}")
```
