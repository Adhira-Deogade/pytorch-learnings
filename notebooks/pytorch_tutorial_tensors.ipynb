```python
# Let's learn about tensors
```

```python
# Pytorch is a tensor based operation. A tensor is a multi-dimensional matrix
# with elements of same datatype
```

```python
import torch
```

```python
x = torch.empty(1)
print(f"empty(1) tensor: {x}")
```

```python
x = torch.empty(3)
print(f"empty(3) tensor: {x}")
```

```python
x = torch.empty(2,4)
print(f"empty(2,4) tensor: {x}")
# 2 arrays of 4 elements
# A matrix of 2 rows, and 4 columns
```

```python
x = torch.empty(1,1)
print(f"empty(1,1) tensor: {x}")
# 1 array of an array with 1 element
# A matrix of 1 row, and 1 column
```

```python
x = torch.empty(1,2)
print(f"empty(1,2) tensor: {x}")
# 1 array of an array with 2 elements
# A matrix of 1 row and 2 columns
```

```python
x = torch.rand(1,2)
print(f"rand(1,2) tensor: {x}")
# 1 array of an array with 2 elements
# A matrix of 1 row and 2 columns
```

```python
# I see the difference between empty and rand
# empty returns values which are almost zero, what computer returns
# rand returns actual values which are non-zero
```

```python
x = torch.rand(2,5)
print(f"rand(2,5) tensor: {x}")
# 2 arrays of 5 elements each
# A matrix of 2 row and 5 columns
```

```python
# scalar = returns an array
# vector/ tensor = returns an array of array
```

```python
x = torch.zeros(2,5)
print(f"zeros(2,5) tensor: {x}")
# 2 arrays of 5 elements each
# A matrix of 2 row and 5 columns
```

```python
# Now I am getting an exact value of zero
```

```python
x = torch.zeros(1)
print(f"zeros(1) tensor: {x}")
# an array of element
```

```python
# We can create not just a matrix but a multi-dimensional matrix
x = torch.rand(1,1,1)
print(f"rand(1,1,1) tensor: {x}")
# An array of an array of an arrays of 1 element
# x, y and z
```

```python
# We can create not just a matrix but a multi-dimensional matrix
x = torch.rand(5,3,4)
print(f"rand(5,3,4) tensor: {x}")
# Create an array - this is always there in tensor
# Create 5 arrays
# Add a 3x4 matrix to each array
# 5 arrays of 3x4 matrices
# An array of matrices is 3 dimensional
```

```python
# We can create not just a matrix but a multi-dimensional matrix
x = torch.rand(5,3,4,2)
print(f"rand(5,3,4,2) tensor: {x}")
# Create an array - this is always there in tensor
# Create 5 arrays
# Add 3 arrays to each of them which are of the size 4x2 matrices
```

```python
# Get size of a tensor
# 1. calling function - size()
# 2. attribute - shape
```

```python
print(f"x.size() = {x.size()}")
print(f"x.shape = {x.shape}")
```

```python
print(f"x.size(0) = {x.size(0)}")
print(f"x.shape[0] = {x.shape[0]}")
```

```python
# Construct tensor from data
x = torch.tensor(
    [
        [1,2]
    ]
)
print(x.dtype)
print(x.shape)
print(x)
```

```python
# Construct tensor from data
x = torch.tensor(
    [
      [1,2],
      [2,3],
      [2,3]
    ]
)
print(x.dtype)
print(x.shape)
print(x)
```

```python
# Construct tensor from data
x = torch.tensor(
    [
        [
          [1,2],
          [2,3],
          [2,3]
        ]
    ]
)
print(x.dtype)
print(x.shape)
print(x)
```

```python
# Construct tensor from data
x = torch.tensor(
    [
        [
          [1,2],
          [2,3],
          [2,3]
        ],
        [
          [1,2],
          [2,3],
          [2,3]
        ]
    ]
)
print(x.dtype)
print(x.shape)
print(x)
```

```python
# Construct tensor from data
x = torch.tensor(
    [
        [
          [1.5,2],
          [2,3],
          [2,3]
        ],
        [
          [1,2],
          [2,3],
          [2,3]
        ]
    ]
)
print(x.dtype)
print(x.shape)
print(x)
```

```python
# If autograd should record operations on the returned tensor
# Calculate gradient for this tensor
# This is a variablel in your model that you want to optimize
x = torch.tensor(
    [
        [
          [1.5,2],
          [2,3],
          [2,3]
        ],
        [
          [1,2],
          [2,3],
          [2,3]
        ]
    ], requires_grad=True,
)
print(x.dtype)
print(x.shape)
print(x)
```

```python
# Mathematical operations on tensors
x = torch.ones([1,2])
y = torch.ones([1,2])
print(x + y)
print(torch.add(x,y))
print(x.add_(y))
```

```python
# Subtraction
x = torch.rand([1,2])
y = torch.rand([1,2])
print(x, y)
print(x-y)
print(x.subtract(y))
print(x.sub_(y))
```

```python
print(x, y)
```

```python
# Multiplication
x = torch.ones([1,2])
y = torch.rand([1,2])
print(x, y)
print(x*y)
print(x.mul(y))
print(x.mul_(y))
print(x, y)
```

```python
# Division
x = torch.ones([1,2])
y = torch.rand([1,2])
print(x, y)
print(x/y)
print(x.div(y))
print(x.div_(y))
print(x, y)
```

```python
# Slicing
x = torch.rand(5,3)
print(x.shape)
print(x)
y = x[:,0] # All rows, 0th column
print(y, y.shape)
z = x[1,:] # First row, all columns
print(z, z.shape)
w = x[0,2] # Get element at 0,2
print(w, w.shape)
```

```python
# From numpy to torch tensor
```

```python
import numpy as np
a = np.ones(5)
print(a, type(a))
b = torch.from_numpy(a) # Will share the same memory
print(b, type(b))
c = torch.tensor(a) # This creates a copy
print(c, type(c))
```

```python
a += 1
print(a)
print(b)
print(c)
```

```python
# GPU support
```

```python
# By default, all the tensors are created on CPU
# We can move them to GPU or create them on GPU
```

```python
device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)
print(device)
```

```python
# If you have multiple GPUs, use -
device = torch.device(
    "cuda:1" if torch.cuda.is_available() else "cpu"
)
print(device)
```

```python
x = torch.rand(5,3).to(device) # Move tensor to GPU device, change datatype also
print(x)
```

```python
print(type(x))
```

```python
print(x.dtype)
```

```python
x = x.to("cuda")
print(x)
```

```python
y = x.to("cpu")
print(y)
```

```python
z = torch.rand(5,3, device=device)
print(z)
```
