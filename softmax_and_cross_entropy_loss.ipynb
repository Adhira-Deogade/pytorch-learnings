```python
def softmax(x):
  return (np.exp(x))/(np.sum(np.exp(x), axis=0))
```

```python
# Example of target with class indices
print("Example of target with class indices")
loss = nn.CrossEntropyLoss()
input = torch.randn(3, 5, requires_grad=True)
print(f"input = {input}")
target = torch.empty(3, dtype=torch.long).random_(5)
print(f"target = {target}")
print(f"target.shape = {target.shape}")
output = loss(input, target)
print(f"output = {output}")
output.backward()
print(f"output = {output}")
```

```python
# Example of target with class probabilities
print("Example of target with class probabilities")
loss = nn.CrossEntropyLoss()
input = torch.randn(3, 5, requires_grad=True)
print(f"input = {input}")
target = torch.randn(3, 5).softmax(dim=1)
print(f"target = {target}")
output = loss(input, target)
print(f"output = {output}")
output.backward()
print(f"output = {output}")
```

```python
def cross_entroy_loss(y, y_pred):
  return -(np.sum(y * np.log(y_pred)))
```

```python
y = np.array([1, 0, 0])
y_pred_good = np.array([0.7, 0.2, 0.1])
y_pred_bad = np.array([0.1, 0.3, 0.6])
print(f"y = {y}")
print(f"y_pred_good = {y_pred_good}")
print(f"y_pred_bad = {y_pred_bad}")
```

```python
cel = cross_entroy_loss(y, y_pred_good)
print(f"cel = {cel}")
```

```python
cel_bad = cross_entroy_loss(y, y_pred_bad)
print(f"cel_bad = {cel_bad}")
```
