```python
from langchain_community.document_loaders import WebBaseLoader

# loader = WebBaseLoader("https://www.example.com/")
```

```python
URL = ["https://www.nytimes.com/2025/03/18/opinion/putin-trump-ukraine-ceasefire.html",
       "https://www.nytimes.com/live/2025/03/17/world/israel-gaza-airstrikes"
]
```

```python
# loader = WebBaseLoader(["https://www.example.com/", "https://google.com"])
loader = WebBaseLoader(URL)
loader.requests_per_second = 1
docs = loader.aload()
docs
```

```python
all_docs = []
docs_page_content = []
docs_page_metadata = []

loader = WebBaseLoader(URL)
loader.requests_per_second = 1

for doc in loader.lazy_load():
    all_docs.append(doc)
    docs_page_content.append(doc.page_content)
    docs_page_metadata.append(doc.metadata)
```

```python
# print(docs_page_content[0][:100])
# print(docs_page_content[1][:100])
print(docs_page_metadata[0])
print(docs_page_metadata[1])
```

```python
all_data = {}
all_data["text"] = docs_page_content
all_data["title"] = docs_page_metadata[0]['title']
all_data["url"] = docs_page_metadata[0]['source']

# Get source from the URL
all_data["source"] = docs_page_metadata[0]['source'].split('.com')[0].split('www.')[-1]
for k, v in all_data.items():
    print(f"k = {k}\nv = {v}")
# Function will take a url and date and then it scrapes data from the URL with webbaseloader and inserts into the database
```
